5. Encuentra el número máximo y mínimo en una lista.

````python
numeros=[7,5,3,1,2,4,6,8]

# Check if the list is not empty to avoid IndexError on initialization
if not numeros:
  print("La lista está vacía.")
else:
  # Initialize aux with the first element for both min (aux[0]) and max (aux[1])
  aux=[numeros[0], numeros[0]]

  for i in range(1, len(numeros)): # Start iterating from the second element
    if numeros[i] > aux[1]: # If current number is greater than current max
      aux[1] = numeros[i]
    elif numeros[i] < aux[0]: # If current number is less than current min
      aux[0] = numeros[i]

aux
````
